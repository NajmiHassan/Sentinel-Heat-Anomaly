```python
import json, pandas as pd
from shapely.geometry import shape

r = json.loads(open("data/heatmaps/day_2024-07-08.json").read())
print(r["result"]["stats_data"])

rows = []
for f in r["result"]["map_data"]["features"]:
    p = f["properties"]; c = shape(f["geometry"]).centroid
    rows.append({"lon": c.x, "lat": c.y, "avg_c": p.get("average_temperature"),
                 "max_c": p.get("max_temperature")})
df = pd.DataFrame(rows)
print(f"tiles: {len(df)}")
print(f"avg spread: {df.avg_c.max() - df.avg_c.min():.2f} °C")
print(f"max spread: {df.max_c.max() - df.max_c.min():.2f} °C")
```

```python
print(df.max_c.describe())
print(df.avg_c.describe())
```

```python
THRESHOLD = round(df.max_c.quantile(0.75), 1)
print(f"use threshold={THRESHOLD}")
```

```python
import json, pathlib
from shapely.geometry import shape

for p in sorted(pathlib.Path("data/heatmaps").glob("day_*.json")):
    feats = json.loads(p.read_text())["result"]["map_data"]["features"]
    peaks = [f["properties"]["max_temperature"] for f in feats]
    print(p.stem[4:], f"peak median {sorted(peaks)[len(peaks)//2]:.1f}  hottest tile {max(peaks):.1f}")
```
